```python
import json
import pandas as pd

# 读取 JSON 文件
with open("data/investors_final.json", encoding="utf-8") as f:
    raw = json.load(f)

# 自动检测 JSON 结构：列表 or 字典
if isinstance(raw, list):
    df = pd.DataFrame(raw)
elif isinstance(raw, dict):
    # 尝试常见的包装键
    for key in ("data", "investors", "records", "results"):
        if key in raw:
            df = pd.DataFrame(raw[key])
            break
    else:
        df = pd.DataFrame([raw])
else:
    raise ValueError(f"不支持的 JSON 结构: {type(raw)}")

print(f"✅ 数据加载完成，共 {len(df):,} 条记录，{df.shape[1]} 列")
```

## 2. 缺失值统计

```python
# 每列缺失值数量和占比
missing = pd.DataFrame({
    "缺失数量": df.isnull().sum(),
    "缺失占比(%)": (df.isnull().mean() * 100).round(1),
})

# 只显示有缺失的列，按缺失量降序排列
missing_only = missing[missing["缺失数量"] > 0].sort_values("缺失数量", ascending=False)

print(f"共 {len(missing_only)} 列存在缺失值（总列数 {df.shape[1]}）\n")
missing_only
```

## 3. 字符串列：唯一值数量 ／ 数字列：基本统计

```python
# 按列类型分开处理
str_cols = df.select_dtypes(include="object").columns.tolist()
num_cols = df.select_dtypes(include="number").columns.tolist()

# ── 字符串列：唯一值数量 + Top-5 高频值 ──────────────────────────────────
print(f"══ 字符串列（共 {len(str_cols)} 列）══\n")
str_summary = []
for col in str_cols:
    n_unique = df[col].nunique()
    top_vals = df[col].value_counts().head(5).index.tolist()
    str_summary.append({
        "列名": col,
        "唯一值数": n_unique,
        "Top-5 高频值": ", ".join(str(v) for v in top_vals),
    })

str_df = pd.DataFrame(str_summary).set_index("列名")
print(str_df.to_string())

print()
```

```python
# ── 数字列：mean / min / max / 非空数量 ──────────────────────────────────
print(f"══ 数字列（共 {len(num_cols)} 列）══\n")
if num_cols:
    num_stats = df[num_cols].agg(["count", "mean", "min", "max"]).T
    num_stats.columns = ["非空数量", "均值", "最小值", "最大值"]
    num_stats["均值"]  = num_stats["均值"].round(2)
    print(num_stats.to_string())
else:
    print("（无数字列）")
```
